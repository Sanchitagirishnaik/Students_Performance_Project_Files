```python
# Calculate total missing values

total_missing = df.isnull().sum().sum()

print("Total Missing Values:", total_missing)
```

```python
# Calculate missing values percentage

missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage)
```

```python
# Create a copy for data cleaning

cleaned_df = df.copy()

print("Copy created successfully!")

cleaned_df
```

```python
# Fill missing Age using Mean

cleaned_df["Age"] = cleaned_df["Age"].fillna(
    cleaned_df["Age"].mean()
)

print("Missing Age filled successfully!")
```

```python
# Fill missing Attendance using Mean

cleaned_df["Attendance"] = cleaned_df["Attendance"].fillna(
    cleaned_df["Attendance"].mean()
)

print("Missing Attendance filled successfully!")
```

```python
# Fill missing Marks using Median

cleaned_df["Marks"] = cleaned_df["Marks"].fillna(
    cleaned_df["Marks"].median()
)

print("Missing Marks filled successfully!")
```

```python
# Fill missing Study Hours using Mean

cleaned_df["Study_Hours"] = cleaned_df["Study_Hours"].fillna(
    cleaned_df["Study_Hours"].mean()
)

print("Missing Study Hours filled successfully!")
```

```python
# Check missing values after cleaning

print("Missing values after cleaning:")

print(cleaned_df.isnull().sum())
```

```python
# Display cleaned dataset

cleaned_df
```

```python
# Save cleaned dataset

cleaned_df.to_csv("cleaned_student_data.csv", index=False)

print("Cleaned dataset saved successfully!")
```

```python
# Remove extra spaces from student names

cleaned_df["Student_Name"] = cleaned_df["Student_Name"].str.strip()

print("Extra spaces removed successfully!")

cleaned_df["Student_Name"]
```

```python
# Convert student names into lowercase

cleaned_df["Student_Name"] = cleaned_df["Student_Name"].str.lower()

print("Student names converted into lowercase!")

cleaned_df["Student_Name"]
```

```python
# Convert department names into lowercase

cleaned_df["Department"] = cleaned_df["Department"].str.lower()

print("Department names converted into lowercase!")

cleaned_df["Department"]
```

```python
# Display cleaned student names and departments

print(cleaned_df[["Student_Name", "Department"]])
```

```python
# Save updated cleaned dataset

cleaned_df.to_csv("cleaned_student_data.csv", index=False)

print("Text cleaned dataset saved successfully!")
```

```python
# Display student marks

print("Student Marks:")

print(cleaned_df["Marks"])
```

```python
# Find minimum and maximum marks

print("Minimum Marks:", cleaned_df["Marks"].min())

print("Maximum Marks:", cleaned_df["Marks"].max())
```

```python
# Calculate IQR for student marks

Q1 = cleaned_df["Marks"].quantile(0.25)

Q3 = cleaned_df["Marks"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR

upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
# Identify outliers in student marks

outliers = cleaned_df[
    (cleaned_df["Marks"] < lower_limit) |
    (cleaned_df["Marks"] > upper_limit)
]

print("Identified Outliers:")

print(outliers[["Student_Name", "Marks"]])
```

```python
# Check the maximum marks after cleaning

print("Maximum Marks After Cleaning:",
      cleaned_df["Marks"].max())

print("Missing Marks:",
      cleaned_df["Marks"].isnull().sum())
```

```python
# Save the final cleaned dataset

cleaned_df.to_csv("cleaned_student_data.csv", index=False)

print("Final cleaned dataset saved successfully!")
```

```python
# Import visualization libraries

import matplotlib.pyplot as plt
import seaborn as sns

print("Matplotlib and Seaborn imported successfully!")
```

```python
# Bar Chart - Student Marks

plt.figure(figsize=(10, 5))

plt.bar(cleaned_df["Student_Name"], cleaned_df["Marks"])

plt.title("Student Performance - Marks")
plt.xlabel("Student Names")
plt.ylabel("Marks")

plt.xticks(rotation=45)

plt.show()
```

```python
# Pie Chart - Department Distribution

department_count = cleaned_df["Department"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    department_count,
    labels=department_count.index,
    autopct="%1.1f%%"
)

plt.title("Student Department Distribution")

plt.show()
```

```python
# Histogram - Marks Distribution

plt.figure(figsize=(8, 5))

sns.histplot(
    data=cleaned_df,
    x="Marks",
    bins=5,
    kde=True
)

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
# Scatter Plot - Attendance vs Marks

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=cleaned_df,
    x="Attendance",
    y="Marks"
)

plt.title("Attendance vs Student Marks")
plt.xlabel("Attendance")
plt.ylabel("Marks")

plt.show()
```
